```python
# Creating sample data 
import pandas as pd
student = {
    'name': ['Arunesh', 'Ramya', 'Mauri', 'Rithvik', 'Siddhanth', 'Anjali'],
    'age': [25, 25, 26, 22, 23, 25],
    'city': ['Perth', 'Bangalore', 'Mumbai', 'Chennai', 'Chennai', 'Sydney'],
    'course': ['MBAn', 'MBA', 'Sport Mgmt', 'MBA', 'BBA', 'MBIS']
}
df = pd.DataFrame(student)
print(df)
                
```

```python
# To View Data
print(df.head())        # First 5 rows
print('\n',df.head(3))
print('\n',df.tail())   # Last 5 rows
print('\n',df.tail(3))
```

```python
#To View Basic Information about the DataFrame
print(df.info())           # Data types, non-null values, memory usage
print('\n',df.describe())  # Statistical summary of numerical columns
print('\n',df.columns)     
print('\n',df.shape)
```

```python
# Accessing Columns
names = df['name'] # single column
age = df['age']
print(names)
print('\n',age)

multiple = df[['name', 'course']] # multiple columns
print('\n',multiple)
```

```python
# Access rows by index
first_row = df.loc[0]        # First row
print(first_row)

# Access Rows by position
first_row = df.iloc[0]       # Same as above
second_third = df.iloc[1:3]  # Rows 1 and 2
print('\n',second_third)

```

```python
# Access specific value
name_at_0 = df.loc[0, 'name']  # Row 0, column "name"
print(name_at_0)  

# Filter Data (Boolean indexing)
high_scorers = df[df['age'] > 23]
print('\n',high_scorers)

chennai_students = df[df['city'] == "Chennai"]
print('\n',chennai_students)

# Multiple conditions (use & for AND, | for OR)
mba_student = df[(df['age'] < 25) & (df['course'] == 'MBA')]
print('\n',mba_student)
```
