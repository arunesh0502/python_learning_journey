```python
print('='*20, 'Loan Approval System', '='*20)

#Applicant Details
name=input('\nEnter Applicant Name: ')
age=int(input('Enter Age: '))
an_in=float(input('Enter Income: $'))
cred=int(input('Enter Credit Score(300-800): '))
years_emp=int(input('Number of Years Employed: '))
has_loans=input('Do You Have Other Loans(yes/no): ').lower()=="yes"
print('-'*60)

#Eligilibilty Criteria
min_age=21
min_in=30000
min_cred=600
min_years=2

#DTI Ratio
loan_amt=10000
mon_pymt=loan_amt/60 #5 Year Plan
mon_in=an_in/12
dtir=(mon_pymt/mon_in)*100

#Check Requirements
age_ok=age>=min_age
income_ok=an_in>=min_in
cred_ok=cred>=min_cred
years_ok=years_emp>=min_years
dti_ok=dtir<=40

#Approval Score Calculation
approval_score=0
if age_ok:
    approval_score+=20
if income_ok:
    approval_score+=25
if cred_ok:
    approval_score+=30
if years_ok:
    approval_score+=15
if dti_ok:
    approval_score+=10
is_approved=approval_score>=80 and not (cred<600)

#Display Results
print('='*60)
print('Loan Approval Decision'.center(60))
print('='*60)
print(f'Applicant Name: {name}')
print(f'Requested Amount: ${loan_amt:.2f}')
print('-'*60)
print(f'Age({min_age}+):                    {age}           -{'OK'if age_ok else 'NO'}')
print(f'Annual Income(${min_in}+):      {an_in}      -{'OK'if income_ok else 'NO'}')
print(f'Credit Score({min_cred}+):          {cred}          -{'OK'if cred_ok else 'NO'}')
print(f'Years Employed({min_years}+):          {years_emp}            -{'OK'if years_ok else 'NO'}')
print(f'Debt To Income Ratio(<40%):  {dtir:.2f}         -{'OK'if dti_ok else 'NO'}')
print('-'*60)
print(f'Approval Score:   {approval_score}/100')
print('-'*60)

if is_approved:
    print('Status Approved')
    print(f'Monthly Payment: ${mon_pymt:.2f}')
    print(f'Interest Rate:   {5.5 if cred>=700 else 7.5}%')
else:
    print('Status Denied')
    print("Improvement Recommendations")
    if not age_ok:
        print(f'Applicant Must Be {min_age} Years Old')
    if not income_ok:
        print(f'Annual Income Should Be Atleast ${min_in}')
    if not cred_ok:
        print(f'Improve Credit Score to Atleast {min_cred}')
    if not years_ok:
        print(f'Should Be Employed For Atleast {min_years}')
    if not dti_ok:
        print(f'Debt To Income Ratio Must Be Less Than 40%')
print('='*60)
```
