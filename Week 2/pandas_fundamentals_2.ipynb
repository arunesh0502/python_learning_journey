```python
# Adding and Modifying Data
import pandas as pd
students = {
    'name':['Arun', 'Mauri', 'Ramya'],
    'age':[25, 26, 25],
    'score':[95, 85, 90],
}
df = pd.DataFrame(students)

# Adding a new column, calculated column and a column based on a condition 
df['grade'] = ['A', 'B', 'A'] 
df['score_doubled'] =  df['score'] * 2
df['passed'] = df['score'] >= 90
print(df)

# Modifying existing data
df.loc[2, 'score'] = 92
df['score_doubled'] = df['score'] * 2
print('\n', df)
```

```python
# Add new row
new_student = {
    'name': 'Rithvik',
    'age': 24,
    'score': 88,
    'grade': 'B',
    'score_doubled': 176,
    'passed': False
}

df = pd.concat([df, pd.DataFrame([new_student])], ignore_index=True)
print(df)
```

```python
# Sorting and Ranking
players = {
    'name': ['Arun', 'Ramya', 'Mauri', 'Rithvik', 'Sid', 'Anjali'],
    'age': [25, 25, 26, 22, 23, 26],
    'score': [95, 90, 85, 88, 92, 89],
}
df = pd.DataFrame(players)


# Sort by score (ascending)
df_sorted = df.sort_values("score")
print(df_sorted)

# Sort by score (descending)
df_sorted = df.sort_values("score", ascending=False)
print('\n', df_sorted)

# Sort by multiple columns
df_sorted = df.sort_values(["age", "score"], ascending=[True, False])
print('\n', df_sorted)

# Add rank column
df["rank"] = df["score"].rank(ascending=False)
print('\n', df)
```

```python
# Aggregation and Grouping
df['city'] = ['Perth', 'Perth', 'Mumbai', 'Chennai', 'Chennai', 'Sydney']

# Basic Statistics
print('Average Score:', f'{df['score'].mean():.2f}')
print('\n', 'Total Score:', df['score'].sum())
print('\n', 'Maximum Score:', df['score'].max())
print('\n', 'Minimum Score:', df['score'].min())
print('\n', 'Standard Deviation:', f'{df['score'].std():.2f}')

# Group by city 
city_stats = df.groupby('city')['score'].mean()
print('\n', city_stats)

# Multiple aggregations
city_summary = df.groupby('city').agg({
    'score': ['mean', 'sum', 'max', 'min'],
    'age': 'mean'
})
print('\n', city_summary)
```
